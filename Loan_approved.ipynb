# EDA

# Univariate analysis

```python
df.head()
```

```python
sns.countplot(x=df.Gender)
```

```python
categorical=[]

for i in df:
    if len(df[i].unique())<=20:
        categorical.append(i)
```

```python
categorical
```

```python
df_cat=df[categorical]
df_cat.head()
```

```python
continuous=[]

for i in df:
    if len(df[i].unique())>=20:
        continuous.append(i)
```

```python
continuous
```

```python
df_cont=df[continuous]
df_cont.head()
```

```python
df_cont = df_cont.drop(columns='Loan_ID')
df_cont.head()
```

```python
plotnumber= 1

plt.figure(figsize=(10,10))
for i in df_cat:
    plt.subplot(3,3,plotnumber)
    sns.countplot(x=df_cat[i])
    plt.xticks(rotation=45)
    plotnumber=plotnumber+1
plt.tight_layout()
```

```python
plotnumber= 1

plt.figure(figsize=(10,4))
for i in df_cont:
    plt.subplot(3,3,plotnumber)
    sns.histplot(x=df_cont[i])
    plt.xticks(rotation=45)
    plotnumber=plotnumber+1
plt.tight_layout()
```

# Bivariate Analysis

# data cleaning

### handling with null values

```python
df.isnull().sum()
```

```python
df.loc[df.Gender.isnull()]
```

```python
df.Gender.value_counts()
```

```python
df.loc[df.Gender.isnull(),'Gender']
```

```python
df.loc[df.Gender.isnull(),'Gender']='Male'
```

```python
df.Gender.isnull().sum()
```

```python
df.loc[df.Married.isnull(),'Married']
```

```python
df.loc[df.Married.isnull(),'Married']='Yes'
```

```python
df.Married.isnull().sum()
```

```python
df.Dependents.value_counts()
```

```python
df.loc[df.Dependents.isnull(),'Dependents']='0'
```

```python
df.Dependents.isnull().sum()
```

```python
df.Self_Employed.value_counts()
```

```python
df.loc[df.Self_Employed.isnull(), 'Self_Employed'] = 'No'
```

```python
df.Self_Employed.isnull().sum()
```

```python
df.loc[df.LoanAmount.isnull()]
```

```python
df.LoanAmount
```

```python
np.median(df.LoanAmount.dropna())
```

```python
df.loc[df.LoanAmount.isnull(), 'LoanAmount'] = np.median(df.LoanAmount.dropna())
```

```python
df.Loan_Amount_Term
```

```python
df.Loan_Amount_Term.value_counts()
```

```python
df.loc[df.Loan_Amount_Term.isnull(), 'Loan_Amount_Term'] = 360.0
```

```python
df.Credit_History.value_counts()
```

```python
df.loc[df.Credit_History.isnull(), 'Credit_History'] = 1.0
```

```python
df.isnull().sum()
```
